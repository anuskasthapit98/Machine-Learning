**Part 1   Regular expressions**

```python
import re

text = """Call Dog Rescue on 0412 345 678 or 0498 111 222.
Your receipt number is 1234 and the date was 14/03/2026 anh.nguyen@spi.nsw.edu.au."""

patterns = {
    "a": r"[Cc]at",          # the word cat or Cat
    "b": r"[0-9]",          # any single digit
    "c": r"\d{4}",          # exactly three digits
    "d": r"\d{2}/\d{2}/\d{4}",          # a date dd/mm/yyyy
    "e": r"04\d{2} \d{3} \d{3}",          # an Australian mobile number
    "f": r"[\w\.-]+@[\w\.-]+\.[a-zA-Z]{2,}",          # email address

}

for name, pattern in patterns.items():
    if pattern:
        print(name, "->", re.findall(pattern, text))
```

**Text Normalisation with NLTK**

```python
text = "The CATS were running quickly, and they didn't stop!"
```

**Task 2.1   Tokenise**

```python
from nltk.tokenize import word_tokenize
tokens = word_tokenize(text)
print(tokens)
```

**Task 2.2   Lowercase and remove punctuation**

```python
import string

lower_tokens = [t.lower() for t in tokens]
clean_tokens = [t for t in lower_tokens if t not in string.punctuation]
print(clean_tokens)
```

**Task 2.3   Stem the tokens**

```python
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()
stemmed = [stemmer.stem(t) for t in clean_tokens]
print(stemmed)
```

**Task 2.4   Put it together**

```python
import string
from nltk.tokenize import word_tokenize
from nltk.stem import PorterStemmer

stemmer = PorterStemmer()

def normalise_text(text):
    tokens = word_tokenize(text)               # TODO 1: split into tokens
    # TODO 2: lowercase, drop punctuation
    tokens = [t.lower() for t in tokens if t not in string.punctuation]
    tokens = [stemmer.stem(t) for t in tokens]      # TODO 3: stem each token
    return tokens

result = normalise_text("Natural Language Processing is FASCINATING!")
Tokens = len(result)
Types = len(set(result))
print(result)
print("Tokens:", Tokens)
print("Types:", Types	)
```
